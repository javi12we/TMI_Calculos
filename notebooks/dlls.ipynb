```python
import pefile

pe = pefile.PE(dll_path)

print("Tipo de DLL:", hex(pe.FILE_HEADER.Machine))
print("Subsystem:", pe.OPTIONAL_HEADER.Subsystem)
print("Imports:")
for entry in pe.DIRECTORY_ENTRY_IMPORT:
    print(" -", entry.dll.decode())
```
